# Local Big Wins Runner

This notebook is for running the local big-win diagnostics from inside VS Code when the terminal runner is unavailable.

```python
from pathlib import Path
import json
import os
import subprocess
import sys

PROJECT = Path('/Users/s0n0611/Documents/GitHub/SLM_PAPER/circuit_breakdown')
STATUS_DIR = PROJECT / 'results/local_big_wins_20260908'
STATUS_DIR.mkdir(parents=True, exist_ok=True)
check = {
    'cwd': os.getcwd(),
    'project_exists': PROJECT.is_dir(),
    'python': sys.executable,
}
(STATUS_DIR / 'notebook_write_check.json').write_text(json.dumps(check, indent=2))
check
```

```python
env = os.environ.copy()
env['PYTHONPATH'] = str(PROJECT / 'src')
env['HF_HUB_OFFLINE'] = '1'
env['TRANSFORMERS_OFFLINE'] = '1'
cmd = ['bash', str(PROJECT / 'local_big_wins/run_local_big_wins.sh')]
process = subprocess.run(cmd, cwd=PROJECT, env=env, text=True, stdout=subprocess.PIPE, stderr=subprocess.STDOUT)
(STATUS_DIR / 'notebook_runner_output.log').write_text(process.stdout)
print(process.stdout[-8000:])
raise SystemExit(process.returncode) if process.returncode else None
```
